### Exercise 3: Swap two values

`Make`

**Quick review**

When you overwrite a variable its old value is gone. To swap two values you need somewhere to keep one of them temporarily.

**Task**

`x` is 3 and `y` is 8. Swap them using a third variable called `temp`, so that `x` becomes 8 and `y` becomes 3.

````python
x = 3
y = 8
temp = x
x = y
y = temp
````

````python
check(x == 8 and y == 3, "After the swap, x must be 8 and y must be 3.")
````

### Exercise 4: Text is not a number

`Fix`

**Quick review**

Every value has a type: `int` (3), `float` (3.5), `str` ("3") and `bool` (`True`). `type(value)` tells you which one. You cannot add text to a number. Convert with `int()`, `float()` or `str()`.

**Task**

This code raises a `TypeError`. Fix it so `total_fruit` is the number 7 (the number of apples plus oranges).

````python
apples = "3"
oranges = 4
total_fruit = int(apples) + oranges
````

````python
check(total_fruit == 7 and isinstance(total_fruit, int), "total_fruit must be the integer 7.")
````

### Exercise 5: Convert and calculate

`Make`

**Quick review**

Data often arrives as text (from files, from `input()`). Convert it before doing maths: `float("19.99")` gives the number `19.99`.

**Task**

`price_text` is text and `quantity` is a number. Store the total price (a number) in `total_price`.

````python
price_text = "19.99"
quantity = 3
total_price = float(price_text) * quantity
````

## Recap

- `=` calculates the right side first and stores the result under the name on the left.
- `b = a` copies the value; later changes to `a` do not affect `b`.
- Tracing code (writing down each variable after each line) is the best way to understand it.
- Types: `int`, `float`, `str`, `bool`. Convert with `int()`, `float()`, `str()`.
- Names: letters, digits and `_`; not starting with a digit; use `snake_case`.

### Common mistakes

- Reading `x = x + 1` as an impossible equation instead of an update.
- Believing that `b = a` keeps `b` linked to `a`.
- Doing maths with text that looks like a number (`"3" + 4`).
