````python
# Run this cell first. It defines the helpers used to check your answers.
import contextlib, io


def check(condition, message="Not quite. Re-read the task and try again."):
    assert condition, message
    print("Correct!")


def check_prediction(prediction, code):
    def clean(text):
        return "\n".join(line.rstrip() for line in str(text).strip().splitlines())

    buffer = io.StringIO()
    with contextlib.redirect_stdout(buffer):
        exec(code, {})
    assert clean(prediction) == clean(buffer.getvalue()), (
        "Not quite. Run the code cell above, compare its output with your prediction, and try again."
    )
    print("Correct!")
````

## Recap

- `print()` shows values; quotes make text, no quotes make a calculation or a name.
- `#` starts a comment; Python ignores it.
- Read error messages from the last line up: error type first, then the line number.
- `SyntaxError` = Python cannot understand the code; `NameError` = a name does not exist.

### Common mistakes

- Forgetting a closing quote or parenthesis.
- Mixing up `Total` and `total` (case matters).
- Putting quotes around something you wanted to calculate: `"2 + 3"` is not `5`.
