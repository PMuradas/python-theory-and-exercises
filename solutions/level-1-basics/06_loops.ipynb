### Exercise 2: The accumulator pattern

`Make`

**Quick review**

A very common pattern: start a variable at a neutral value (0 for sums), then update it on every iteration. `total += n` is a short form of `total = total + n`.

**Task**

Add up the numbers from 1 to 100 (both included) and store the result in `total`.

````python
total = 0
for n in range(1, 101):
    total += n
````

````python
check(total == 5050, "The sum of 1 to 100 is 5050.")
````

### Exercise 4: Off by one

`Fix`

**Quick review**

The most common loop bug: forgetting that `range()` excludes its last number.

**Task**

This should add the numbers from 1 to 10 (55) but gives a different result. Fix it.

````python
total = 0
for n in range(1, 11):
    total += n
````

````python
check(total == 55, "The sum of 1 to 10 is 55.")
````

### Exercise 5: Loop with a condition

`Make`

**Quick review**

Put an `if` inside a loop to act only on some values. `or` combines conditions.

**Task**

Add up all the numbers below 100 that are divisible by 3 or by 5. Store the result in `special_sum`.

````python
special_sum = 0
for n in range(1, 100):
    if n % 3 == 0 or n % 5 == 0:
        special_sum += n
````

````python
check(special_sum == 2318, "The answer for numbers below 100 is 2318.")
````

### Exercise 6: Stopping early with break

`Make`

**Quick review**

`break` leaves the loop immediately. `while True:` loops "forever" until a `break` runs. Warning: a `while` loop whose condition never becomes false never stops; if that happens, use the Stop button (or Runtime > Interrupt execution).

**Task**

Find the smallest whole number `n` (starting from 1) whose square is greater than 500, using a `while True` loop and `break`.

````python
n = 1
while True:
    if n * n > 500:
        break
    n += 1
````

````python
check(n == 23, "23 * 23 = 529 is the first square above 500.")
````

### Exercise 7: Nested loops

`Make`

**Quick review**

A loop inside a loop runs the inner loop completely for every step of the outer loop. Great for tables and grids.

**Task**

Add up the products `i * j` for every pair of `i` and `j` from 1 to 9 and store it in `table_sum`.

````python
table_sum = 0
for i in range(1, 10):
    for j in range(1, 10):
        table_sum += i * j
````

````python
check(table_sum == 2025, "The sum of the 9 x 9 multiplication table is 2025.")
````

## Recap

- `range(stop)` starts at 0 and excludes `stop`.
- Accumulator pattern: initialise before the loop, update inside it.
- `while` needs something inside the loop that eventually makes the condition false.
- `break` leaves a loop; `continue` skips to the next iteration.

### Common mistakes

- Off-by-one with `range()`.
- Forgetting to update the variable in a `while` loop (infinite loop).
- Initialising the accumulator inside the loop, which resets it every time.
