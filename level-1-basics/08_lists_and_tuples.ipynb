## Recap

- Lists are changeable and ordered; tuples are not changeable.
- `b = a` does not copy a list: both names point to the same list. Use `.copy()`.
- `sorted()` returns a new list; `.sort()` changes the list in place.
- Build results with the pattern: empty list, loop, `append`.

### Common mistakes

- Expecting `b = a` to make an independent copy.
- Writing `x = my_list.sort()` and getting `None`.
- Index errors from forgetting that the last index is `len(list) - 1`.
