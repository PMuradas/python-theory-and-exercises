## Recap

- Only the first true branch of an `if`/`elif`/`else` chain runs.
- Indentation is part of Python's syntax: it defines the block.
- `==` compares, `=` assigns.
- Empty values (`""`, `0`, empty collections) are falsy; the rest are truthy.

### Common mistakes

- Forgetting the colon after the condition.
- Using `=` instead of `==`.
- Checking a broad condition before a specific one, so the specific branch never runs.
