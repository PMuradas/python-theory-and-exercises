## Recap

- `/` gives a float, `//` rounds down, `%` is the remainder, `**` is the power.
- Parentheses make the order explicit; use them whenever precedence is unclear.
- Comparisons produce booleans; `and`, `or`, `not` combine them.
- `n % 2 == 0` is the classic even check.

### Common mistakes

- Using `=` (assign) when you meant `==` (compare).
- Expecting `7 / 2` to be `3`.
- Writing `-3 ** 2` and expecting `9`.
