## Recap

- `def name(parameters):` defines a function; `return` sends a value back.
- `print` shows, `return` gives back. No `return` means `None`.
- Default values make parameters optional.
- Variables created inside a function are local to it.

### Common mistakes

- Printing inside a function when you needed to return.
- Defining a function but never calling it.
- Trying to use a local variable outside its function.
