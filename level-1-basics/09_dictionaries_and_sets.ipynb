## Recap

- Dictionaries map keys to values; `d[key]` raises `KeyError` if missing, `d.get(key, default)` does not.
- `d.items()` gives (key, value) pairs.
- Counting pattern: `counts[x] = counts.get(x, 0) + 1`.
- Sets keep unique items; `&` and `|` combine them.

### Common mistakes

- Reading a key that does not exist.
- Expecting a set to keep order or duplicates.
- Writing `{}` and expecting an empty set (it is an empty dictionary; use `set()`).
