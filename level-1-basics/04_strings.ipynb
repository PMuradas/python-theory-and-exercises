## Recap

- Indexing starts at 0; negative indexes count from the end.
- `text[start:stop]` includes start and excludes stop.
- String methods return new strings; the original never changes.
- f-strings are the clearest way to build text from values.

### Common mistakes

- Off-by-one with slices (`stop` is excluded).
- Calling `word.upper()` and expecting `word` itself to change.
- Forgetting the `f` before the quote in an f-string.
